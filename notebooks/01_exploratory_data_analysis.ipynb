```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Caminho da pasta com os dados originais
DATA_PATH = Path("../data/raw")

# Listar arquivos CSV disponíveis
files = list(DATA_PATH.glob("*.csv"))

print("Arquivos encontrados:", len(files))

for file in files:
    print(file.name)
```

```python
# Carregar todas as tabelas CSV
dataframes = {}

for file in files:
    name = file.stem.replace("olist_", "").replace("_dataset", "")

    dataframes[name] = pd.read_csv(
        file,
        low_memory=False
    )

print("Tabelas carregadas com sucesso!")
```

```python
# Mostrar dimensões de todas as tabelas

for name, df in dataframes.items():

    print(f"\nTabela: {name}")
    print(f"Linhas: {df.shape[0]}")
    print(f"Colunas: {df.shape[1]}")
```

```python
# Verificar valores ausentes em todas as tabelas

for name, df in dataframes.items():

    missing = df.isnull().sum()

    print(f"\nTabela: {name}")

    print(
        missing[missing > 0].sort_values(ascending=False)
    )
```

```python
# Verificar registros duplicados

for name, df in dataframes.items():

    duplicates = df.duplicated().sum()

    print(f"{name}: {duplicates} registros duplicados")
```

```python
quality_report = []

for name, df in dataframes.items():

    quality_report.append({
        "Tabela": name,
        "Linhas": len(df),
        "Colunas": len(df.columns),
        "Valores ausentes": df.isnull().sum().sum(),
        "Duplicados": df.duplicated().sum()
    })

quality_df = pd.DataFrame(quality_report)

quality_df
```

```python
quality_df.sort_values(
    by="Valores ausentes",
    ascending=False
)
```

```python
quality_df.to_csv(
    "../docs/data_quality_report.csv",
    index=False
)

print("Relatório salvo com sucesso!")
```
